**Cálculos Básicos: Adição, subtração, multiplicação, divisão, potenciação e radiciação**


```python
import math  # estou impotando uma biblioteca de matemática

# inserir os números para o cálculo

num1 = float(input("Digite o primeiro número:\n "))
num2 = float(input("Digite o segundo número:\n"))

# inserir e escolher a operação

operacao = input("Digite a operação desejada (+, -, *, /)  digite (^ ) para potência e ( r ) para raiz quadrada:\n")

# aqui temos as operações

if  operacao == "+":
  resultado = num1 + num2
  print("O seu resultado é:", resultado)


elif operacao == "-":
  resultado = num1 - num2
  print("O seu resultado é:", resultado)


elif operacao == "*":
  resultado = num1 * num2
  print("O seu resultado é:", resultado)


elif operacao == "/":
      resultado = num1 / num2
      print(f"{resultado:.2f}")     # aqui eu estou arrendondando a divisão com o :.2f


elif operacao == "^":
  resultado = num1 ** num2
  print("O seu resultado é:", resultado)


elif operacao == "r":
  resultado = math.sqrt(num1)      # sqrt é a chave para fazer raiz quadrada que eu importei lá em cima
  print(f"{resultado:.2f}")              # aqui é o resultado arredondado
else :
    print("O seu resultado é:", resultado)
```

Com base no cenário de consultoria da Talita, onde o valor cobrado é composto por uma taxa fixa de 380,00 mais 120,00 por hora trabalhada, desenvolva:

**Cálculo de Receita:** Uma função que retorne o valor total a ser cobrado com base nas horas trabalhadas.

**Cálculo Reverso:** Uma função que determine a quantidade de horas trabalhadas a partir de um valor total informado.

```python
#Cálculo da receita

taxa_fixa = 380.00
valor_hora = 120.00
horas_trabalhadas = float(input("Digite a quantidade de horas trabalhadas: \n"))

def calcular_receita():
    total = taxa_fixa+ (valor_hora * horas_trabalhadas)
    return total
print(f"O valor total a ser cobrado é: R$ {calcular_receita():.2f}")
```

```python
#Cálculo Reverso

taxa_fixa = 380.00
valor_hora = 120.00
valor_total= float(input("Digite a quantidade que você recebeu:\n"))

def calcular_horas(valor_total):
    if valor_total < taxa_fixa:
        return 0

    horas = (valor_total - taxa_fixa) / valor_hora
    return horas
print(f"A quantidade de horas trabalhadas é: {calcular_horas(valor_total):.2f} horas")
```

**Geometria Analítica e Álgebra**

Dadas as coordenadas de dois pontos quaisquer,  A(xA, yA) e B(xB, yB),  crie funções para:

Calcular a taxa de variação (coeficiente angular a).

Calcular o intercepto do eixo y (coeficiente linear b).

Exibir a lei de formação da função no formato f(x) = ax + b.


```python
x1= float(input("Digite o valor de x1: \n"))
y1= float(input("Digite o valor de y1: \n"))
x2= float(input("Digite o valor de x2: \n"))
y2= float(input("Digite o valor de y2: \n"))


def calcular_coeficiente_angular(x1, y1, x2, y2):   # Calcula a taxa de variação (a).  Fórmula: (y2 - y1) / (x2 - x1)
    a = (y2 - y1) / (x2 - x1)
    return a
print(f"O coeficiente angular é: {calcular_coeficiente_angular(x1, y1, x2, y2):.2f}")

```
